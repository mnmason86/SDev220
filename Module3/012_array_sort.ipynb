##### Given an array arr[] containing only 0s, 1s, and 2s. Sort the array in ascending order. 

In real life I would just use the .sort() method to sort an array in ascending order:

```python

def sort012(arr):
    arr.sort()
    return arr
    
sorted = sort012([0, 1, 2, 0, 1, 2])
print("sorted array:", sorted)
    
```

But for the sake of showing that I understand how it works:

```python
def merge_sort(arr):
    
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j <len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result


merged_array = merge_sort([0, 1, 2, 0, 1, 2])
print("sorted array:", merged_array)
```
