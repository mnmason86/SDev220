#### 7.5 Capitalize the element in things that refers to a person and then print the list. Did it change the element in the list?

*It did capitalize the element in the list, though I needed to specify which element to modify specifically.*

```python
for i in range(len(things)):
    if things[i] == "cinderella":
        things[i] = things[i].capitalize()
        
print(things)
checkForSalmonella(things)
```

#### 7.6 Make the cheesy element of things all uppercase and then print the list.

```python
for i in range(len(things)):
    if things[i] == "mozzarella":
        things[i] = things[i].upper()
        
print(things)
checkForSalmonella(things)
```

#### 7.7 Delete the disease element from things, collect your Nobel Prize, and print the list.

```python
for i in range(len(things)):
    if things[i] == "salmonella":
        del things[things.index(things[i])]
    
print(things)
checkForSalmonella(things)
```

#### 9.2 Define a generator function called get_odds() that returns the odd numbers from range(10). Use a for loop to find and print the third value returned.

```python
def get_odds():
    for i in range(1, 10):
        if i % 2 == 1:
            yield i
            
for i, val in enumerate(get_odds(), start=1):
    if i == 3:
        print(val)
        break

```
