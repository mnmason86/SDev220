16.4 Use the sqlite3 module to create a SQLite database called books.db and a table called books with these fields: title (text), author (text), and year (integer).

```python
import sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()
curs.execute('''CREATE TABLE IF NOT EXISTS books
                (title TEXT, author TEXT, year INTEGER)''')
```

16.5 Read books2.csv and insert its data into the book table.

```python
with open('books2.csv', 'r') as f:
    next(f)  # skip the header row
    for line in f:
        title, author, year = line.strip().split(',')
        curs.execute("INSERT INTO books (title, author, year) VALUES (?, ?, ?)", (title, author, int(year)))
        print(f"Inserted book: {title} by {author} ({year})")

conn.commit()
```

16.8 Use the sqlalchemy module to connect to the sqlite3 database books.db that you just made in exercise 16.4. As in 16.6, select and print the title column from the book table in alphabetical order.

```python
# %pip install sqlalchemy

import sqlalchemy as sa

engine = sa.create_engine('sqlite:///books.db')
connection = engine.connect()
result = connection.execute(sa.text("SELECT title FROM books ORDER BY title"))
for row in result:
    print(row.title)

# clear the database for the next run
# curs.execute("DELETE FROM books")
```
